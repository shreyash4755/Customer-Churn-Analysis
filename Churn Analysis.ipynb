```python
#Step 1: Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Step 2: Load Data

df = pd.read_csv(r"C:\Users\Shreyash Tilekar\Downloads\WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.head()
```

```python
#Step 3: Explore Data

df.info()
df.describe()
df.isnull().sum()
df['Churn'].value_counts()
```

```python
#STEP 4: Data Cleaning

# Remove spaces and handle missing values
df['TotalCharges'] = df['TotalCharges'].replace(' ', np.nan)
df['TotalCharges'] = df['TotalCharges'].astype(float)
df.dropna(inplace=True)

# Convert 'Churn' to binary
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

df.head()
```

```python
#STEP 5: Exploratory Data Analysis (EDA)

#Churn Distribution
sns.countplot(x='Churn', data=df)
plt.title('Customer Churn Distribution')
plt.show()
```

```python
#Gender vs Churn
sns.countplot(x='gender', hue='Churn', data=df)
plt.title('Churn by Gender')
plt.show()
```

```python
#Contract Type vs Churn
sns.countplot(x='Contract', hue='Churn', data=df)
plt.title('Churn by Contract Type')
plt.show()
```

```python
#Monthly Charges vs Churn
sns.boxplot(x='Churn', y='MonthlyCharges', data=df)
plt.title('Monthly Charges vs Churn')
plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap (Numeric Columns Only)")
plt.show()
```

```python
#Step 6: Key Insights

print("""
✅ Key Insights:
1. Month-to-month contracts have higher churn rates.
2. Customers with high monthly charges are more likely to churn.
3. Longer tenure and annual contracts reduce churn likelihood.
4. Online security and tech support services help retain customers.
""")
```
